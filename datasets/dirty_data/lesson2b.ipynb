```python
import pandas as pd 
import numpy as np
```

```python
df['Age']=pd.to_numeric(df['Age'], errors='coerce')
df
```

```python
df.loc[df['Age'] < 0, 'Age']= np.nan
df
```

```python
df['Age']= df['Age'].fillna(df['Age'].median()).astype(int)
df
```

```python
df['Gender']= df['Gender'].str.strip()
df['Gender']=df['Gender'].str.capitalize()
df['Gender'].unique()
```

```python
df['Gender']=df['Gender'].replace({"M":"Male"})
df['Gender'].value_counts()
```

```python
df['Gender']= df['Gender'].replace(['M', 'F'], ['Male', 'Female'])
df
```

```python
df.loc[df['Study_Hours'] > 24,'Study_Hours']=4.0
df['Study_Hours']= df['Study_Hours'].fillna(3.0)
df
```

```python
df['Attendance']= df['Attendance'].str.replace('%', '')
df['Attendance']= df['Attendance'].astype(float)
df.loc[df['Attendance'] > 100, 'Attendance'] = 100.0
df['Attendance'] = df['Attendance'].fillna(85.0)
df
```
