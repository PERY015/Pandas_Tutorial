```python
df.dtypes.unique()
```

```python
df['name'].unique()
df['name']= df['name'].str.strip()
df['name']= df['name'].str.capitalize()
df['name'].unique()
```

```python
df['age'].value_counts()
df['age']=pd.to_numeric(df['age'],errors='coerce')
df.loc[(df['age']<0) | (df['age']>100)]=np.nan
df.dropna(inplace=True)
df
```

```python

df['gender']= df['gender'].str.strip()
df['gender']= df['gender'].str.capitalize()
df['gender'].unique()
```

```python
df['gender']=df['gender'].replace({
    "M":"Male"
})
df['gender'].value_counts() 
```

```python
df['city']=df['city'].str.strip().str.capitalize()
df
```

```python
email_pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"
valid_email=df['email'].str.match(email_pattern, na=False)
print(valid_email)
df.loc[~valid_email,'email']=pd.NA
```

```python
df
```

```python
df['phone']=df['phone'].str.replace('-','').str.replace(' ','')
df
```

```python
df['phone']=df['phone'].str.replace(r"^07","+2547",regex=True)
df
```

```python
df['signup_date']=df['signup_date'].str.replace('/', '-', regex=False)
df

```

```python
df['signup_date']=df['signup_date'].apply(lambda x:pd.to_datetime(x,errors='coerce'))
df
```

```python
df['purchase_amount']=df['purchase_amount'].astype(str).str.replace(',', '').str.strip()
df
```

```python
df['purchase_amount']=pd.to_numeric(df['purchase_amount'], errors='coerce')
df
```

```python
df['purchase_amount']=df['purchase_amount'].fillna(0)
df
```

```python
df['category']=df['category'].astype(str).str.lower().str.strip()
df
```
